# PyNorma Functionality Showcase
This notebook demonstrates the major functionalities of the `pynorma` library using the provided example files.

## 1. Parsing Data
Pynorma provides functions to parse CSV and XLSX files with automatic trimming and header detection.

```python
import pandas as pd
from pynorma.io import csv_parser, xlsx_parser
from pynorma.io import csv_writer
from pynorma.preprocessor import appender, atomizer, clarifier, merger, flattener

# Set pandas display options
pd.set_option('display.max_rows', 10)
```

### 1.1. Parsing a CSV file
Here we parse `townbusiness1.csv`, which has extra rows and columns around the data.

```python
df_csv = csv_parser.parse_csv('../examples/townbusiness1.csv', verbose=True)
print("Parsed CSV data:")
display(df_csv)
```

### 1.2. Parsing an XLSX file
Here we parse `crimetable.xlsx`, which has merged cells and multiple header rows.

```python
df_xlsx = xlsx_parser.parse_xlsx('../examples/crimetable.xlsx')
print("Parsed XLSX data:")
display(df_xlsx)
```

## 2. Appending DataFrames
We will now append `townbusiness2.csv` to the dataframe we parsed from `townbusiness1.csv`.

```python
df_csv2 = csv_parser.parse_csv('../examples/townbusiness2.csv', isheader=False)
df_csv2.columns = df_csv.columns
df_appended = appender.append(df_csv, df_csv2, verbose=True)
print("Appended DataFrame:")
display(df_appended)
```

## 3. Atomizing Data
The 'Business' column contains multiple values separated by commas. We will use `atomize_by_column` to separate them into individual rows.

```python
df_atomized = atomizer.atomize_by_column(df_appended, atm_cols=['Business'], delimiter=',')
print("Atomized DataFrame:")
display(df_atomized)
```

## 4. Clarifying Data
Now we'll clean up the 'Business' column by correcting typos and standardizing synonyms using a dictionary.

```python
df_clarified = clarifier.clarify(df_atomized, column='Business', dict_path='../examples/townbusiness_dict.csv', sum_columns=['count', 'staff'])
print("Clarified DataFrame:")
display(df_clarified)
```

## 5. Merging Data
After clarifying, we have duplicate rows. We'll merge them and sum the 'count' and 'staff' columns.

```python
df_merged = merger.merge(df_clarified, sum_column=['count', 'staff'])
print("Merged DataFrame:")
display(df_merged)
```

## 6. Flattening Data
This demonstrates how to convert a wide-format table with a multi-level header into a long-format (tidy) table.

```python
df_wide = pd.read_excel('../examples/Kor119.xlsx', header=None)
df_flattened = flattener.flatten(df_wide, id_vars=[0, 1, 2], header_end=2, time_level=2)
print("Flattened DataFrame:")
display(df_flattened)
```

## 7. Writing Data
Finally, we'll save one of our processed DataFrames to a new CSV file.

```python
output_path = '../examples/result/townbusiness_processed.csv'
csv_writer.write_output(df_merged, output_path)
print(f'Successfully wrote processed data to {output_path}')
```
